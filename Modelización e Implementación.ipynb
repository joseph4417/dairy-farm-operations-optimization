# PLAN DE OPERACIONES DE GRANJA BOVINA


Se desea desarrollar un plan de producción para una finca lechera durante un horizonte de planificación de cinco años, considerando variables como venta y conservación de ganado, personal contratado y alimentación , utilizando técnicas de programación lineal entera y sujeto a las siguientes características:

- Actualmente, la granja cuenta con un hato de 95 animales, los cuales se dividen en tres grupos: terneras (cría de 0 a 1 año), vaconas (vaca joven de 1 a 2 años) y vacas lecheras (vacas adultas de más de 2 años). 

- Cada vaca, de acuerdo a la categoría a la que pertenezca, requiere de cierta cantidad de terrenos para pastar. En la actualidad, la granja tiene una extensión de 25 hectaéreas. 

- Cada vaca lechera produce cierta cantidad de leche al día y es productiva durante 10 meses al año. La leche es comercializada con empresas de
productos lácteos a cierto costo por litro. 

- El ciclo productivo de una vaca lechera es el siguiente:
    
    - Una vaca pare un ternero. 
    - Durante 2 meses la vaca está en periodo de recuperación post - parto. Al finalizar el tercer mes puede volver a gestar. En estos tres primeros meses la vaca sí produce leche. El periodo de gestación de una vaca es de 9 meses. 
    - Una vez la vaca entre en gestación, esta sigue produciendo leche durante los siguientes 7 meses.
    - Con el fin de asegurar el bienestar del feto y la madre, los últimos dos meses de gestación no se realiza el ordeño. 
    - Una vez nacido el ternero se repite el ciclo. 

- Cada vaca puede producir 1 ternero al año. Suponemos que la relación de nacidos machos y hembras es 1:1. 

- Cuando un ternero nace se tienen dos opciones:

    - Si es macho, se vende de forma inmediata pues no es rentable mantenerlo en
la granja.

    - Si es hembra existen dos posibilidades: se puede vender de forma inmediata a
un precio mayor que un macho o puede mantenerse en la finca hasta convertirse
en vaca lechera.

-  La práctica general de la granja es vender una vaca lechera a los 11 año. Además, de forma anual, cierto porcentaje de las vaconas muere debido a
enfermedades o problemas sanitarios y cierto porcentaje de las vacas lecheras salen
del sistema productivo por mortalidad. 

- Respecto a la alimentación del ganado, la granja proporciona una dieta suplementaria fija de la mezcla de cáscara de maracuyá, desechos de piñas, afrecho de
cerveza, pollinaza (desecho de granjas avícolas) y sal mineralizada. Todos los animales consumen la misma mezcla pero en diferentes cantidades de acuerdo a la categoría a la que pertenecen.  Puesto que la
dieta es fija, se puede considerar que el costo de alimentación es un parámetro fijo.

- La granja tiene una capacidad de almacenamiento de alimento limitado, y cada
vaca requiere cierta cantidad de alimento (dieta suplementaria y pastoreo) para
mantenerse en condiciones de productividad.

- Adicionalmente, se requiere ciertas horas de trabajo para las labores de la granja (cuidado de animales, almacenamiento de alimentos y manejo sanitario).  

### Objetivo
Debemos determinar el mejor plan de operación para los próximos cinco años con el objetivo de maximizar la utilidad de la granja, definiendo:

- el número de terneros a vender y conservar, 
- la cantidad de alimento a adquirir, 
- el número de trabajadores fijos requeridos y 
- el número de horas extras constratadas. 

manteniendo en cuenta la capacidad de almacenamiento de alimento, extensión del terreno de pastoreo, requerimientos alimenticios del ganado y mano de obra. 


## MODELIZACIÓN


### CONJUNTOS

- Sea $Y= \{1,2, \dots, n\}$ el conjunto de años, donde $n$ es el horizonte de tiempo. 
- Sea $A= \{0,1,\cdots, m\}$ el conjunto de edades de las vacas, donde $m$ es la edad a la que un animal se destina a venta. Una animal se considera:
    - ternero desde que nace hasta el año 1. 
    - vacona desde el año 1 hasta el año 2. 
    - vaca lechera desde el año 2 en adelante. A este conjunto lo denotamos por $A_l = \{2,\cdots, m\}$.
    - de descarte si su edad es mayor a $m$. 
- Sea $X=\{1,2, \cdots, 12\}$ el conjunto de meses en un año. 

### PARÁMETROS

#### Respecto al ganado: 
- Sea $t_0$ el número inicial de terneras. 
- Sea $v_0$ el número inicial de vaconas. 
- Sea $l_{e,0}$ el número inicial de vacas de edad $e$, $\forall e \in A_l$.
- Sea $\alpha_t \in [0,1]$ la tasa de mortalidad de terneros. 
- Sea $\alpha_v \in [0,1]$ la tasa de mortalidad de vaconas. 
- Sea $\alpha_e \in [0,1]$ la tasa de mortalidad de vacas de edad $e$, $\forall e \in A_l$. 
- Sea $\beta \in [0,1]$ el porcentaje de nacimiento de hembras. Así, $1 - \beta$ es el porcentaje de nacimiento de machos. 
- Sea $\delta \in [0,1]$ el procentaje de nacimientos exitosos.  
- Sea $Q$ el número de ganado mínimo que debemos mantener al final de los cinco años. 

#### Respecto a la producción:
- Sea $\mu$ la cantidad de litros producidos por una vaca durante un día. 
- Sea $d$ el número de días productivos al año. 

#### Respecto a los precios:
- Sea $p_{\mu}$ el precio de venta de un litro de leche. 
- Sea $p_{t_M}$ el precio de venta de un ternero macho. 
- Sea $p_{t_H}$ el precio de venta de un ternero hembra. 
- Sea $p_D$ el precio de venta de una vaca descartada. 

#### Respecto a la alimentación:
- Sea $c_f$ el costo por kilogramo de alimento suplementario. 
- Sea $r_t$ la cantidad en kilogramos de alimento requerido por una ternera. 
- Sea $r_v$ la cantidad en kilogramos de alimento requerido por una vacona. 
- Sea $r_l$ la cantidad en kilogramos de alimento requerido  por una vaca lechera. 
- Sea $b$ el número de bodegas de almacenamiento de alimento.
- Sea $k$ la capacidad máxima de almacenamiento de alimento por mes en una bodega, medida en kilogramos. 
- Sea $c_a$ el costo por almacenamiento por kilogramo de alimento. 

#### Respecto al terreno:
- Sea $H$ la extensión del terreno de la granja, medido en hectareas. 
- Sea $h_t$ el número de hectareas requeridas por una ternera para pastoreo. 
- Sea $h_v$ el número de hectareas requeridas por una vacona para pastoreo. 
- Sea $h_l$ el número de hectareas requeridas por una vaca lechera para pastoreo. 

#### Respecto a la mano de obra:
- Sea $w_t$ las horas de manejo y cuidado (salud, alimentación, adecuación de terreno y saneamiento) requeridas por ternera por año. 
- Sea $w_v$ las horas de manejo y cuidado  (salud, alimentación, adecuación de terreno y saneamiento) requeridas por vacona por año. 
- Sea $w_l$ las horas de manejo y cuidado  (salud, alimentación, adecuación de terreno y saneamiento) requeridas por vaca lechera por año. Aquí se incluye las horas de ordeño. 
- Sea $c_{reg}$ el costo  por hora regular de trabajo. 
- Sea $c_{ext}$ el costo por hora extra de trabajo. 
- Sea $s$ el número de horas que aporta un trabajador al año. 

#### Respecto a los costos operativos:
- Sea $c_t$ el costo anual de manejo y cuidado por ternera. 
- Sea $c_v$ el costo anual de manejo y cuidado por vacona.
- Sea $c_l$ el costo anual de manejo y cuidado por vaca lechera.

Mantener un animal vivo es, en esencia, una decisión de inversión pues se incurre en costos hoy (alimentación, cuidado de una vacona) esperando ingresos futuros (producción de leche, venta al descarte). Por esta razón, incorporaremos flujo de caja al modelo para evaluar este tipo de decisiones. Se considera los siguientes parámetros: 

- Sea $p_{g_T}$ el valor terminal (año $n$) de un animal en transición general hacia el ciclo productivo, antes de convertirse en vacona.
- Sea $p_{g_V}$ el valor terminal (año $n$) de una vacona, correspondiente al valor esperado descontado de que llegue a ser vaca productiva de edad 2. 
- Sea $p_{g_e}$ el valor terminal (año $n$) de una vaca lechera de edad e, correspondiente al valor esperado descontado de sus años productivos futuro, $\forall e \in A_l$.Se fija $p_{g_m} = 0$. 

El modelo proyecta flujos de caja hasta el año $n$, pero un animal vivo en ese punto sigue generando valor económico más allá del horizonte de análisis. Estos valores terminales  se incluyen precisamente para capturar, en un solo número, la utilidad descontada a futuro que representa mantener a ese animal con vida en la granja, en lugar de descartar ese valor remanente al truncar la proyección.
Cada valor terminal incorpora tanto el descuento financiero como la probabilidad de pérdida o descarte del animal según su edad, de modo que refleja el valor esperado real —y no solo teórico— de conservar animales en distintas etapas productivas. El caso $p_{g_m} = 0$ evita doble conteo, ya que ese animal se vende en el mismo año $m$ y su valor ya queda registrado en el flujo de ese período.
En conjunto, estos cálculos permiten que el modelo entregue una estimación de rentabilidad más completa y realista, al reconocer el valor futuro del ganado que permanece en la granja más allá del horizonte explícito de proyección.

### VARIABLES DE DECISIÓN



- Sea $T_y \in \mathbb{N}$ el número de terneras al final del año $y$, $\forall y \in Y$.
- Sea $V_y \in \mathbb{N}$ el número de vaconas al final del año $y$, $\forall y \in Y$.
- Sea $L_{e,y} \in \mathbb{N}$ el número de vacas lecheras de edad $e$ al final del año $y$, $\forall e \in A_l, \forall y \in Y$.
- Sea $M_y \in \mathbb{N}$ el número de terneros vendidos al final del año $y$, $\forall y \in Y$.
- Sea $H_y \in \mathbb{N}$ el número de terneras vendidas al final del año $y$, $\forall y \in Y$.
- Sea $F_{x,y} \geq 0$ los kilogramos de alimento suplementario comprado en el mes $x$ al año $y$, $\forall x \in X, \forall y \in Y$.
- Sea $S_y$ el número de trabajadores fijos contratados durante el año $y$, $\forall y \in Y$.
- Sea $O_y \in \mathbb{N}$ el número de horas extras contratadas en el año $y$, $\forall y \in Y$.

### FUNCIÓN OBJETIVO 

Debemos considerar cada uno de los ingresos y costos que tiene la granja.

Para $y \in Y$ fijo, se tiene:

1. Ingresos por producción de leche:

$$
p_{\mu} \cdot {\mu} \cdot d \cdot \sum_{e \in A_l} L_{e,y}
$$

2. Ingreso por venta de terneros machos: 

$$
p_{t_M} \cdot M_y
$$

3. Ingreso por terneras hembras vendidas:

$$
p_{t_H} \cdot H_y
$$

4. Ingresos por vacas de descarte a vender al final del año:

$$
p_{D} \cdot L_{m,y}
$$

5. Costos de compra de alimento y almacenamiento:

$$
(c_f + c_a)\cdot \sum_{x \in X}F_{x,y}
$$

6. Salarios de trabajadores fijos:

$$
c_{reg} \cdot s \cdot S_y
$$

7. Costo de horas extras contratadas:

$$
c_{ext} \cdot O_y
$$

8. Costos de manejo y cuidado:

$$
c_t \cdot  T_y + c_v \cdot V_y + c_l\sum_{e \in A_l} L_{e,y}
$$

9.  utilidad descontada a futuro:

$$
p_{g_T} \cdot T_n + p_{g_V} \cdot V_n + \sum_{e \in A_l} p_{g_e} \cdot L_{e,n}
$$

Nuestro propósito es maximizar las ganancias y minimizar los costos de la granja en cada año, así, nuestra función objetivo es:

$$
\begin{aligned}
\max \sum_{y \in Y} \Bigl[
& p_{\mu} \cdot \mu \cdot d \cdot \sum_{e \in A_l} L_{ey}
+ p_{t_M} \cdot M_y
+ p_{t_H} \cdot H_y
+ p_D \cdot L_{m,y} \\
& - (c_f + c_a) \cdot \sum_{x \in X} F_{x,y}
- c_{reg} \cdot s \cdot S_y
- c_{ext} \cdot O_y
- \left(c_t \cdot T_y + c_v \cdot V_y + c_l \cdot \sum_{e \in A_l} L_{ey}\right)
\Bigr] \\
& + p_{g_T} \cdot T_n
+ p_{g_V} \cdot V_n
+ \sum_{e \in A_l} p_{g_e} \cdot L_{e,n}.
\end{aligned}
$$

### RESTRICCIONES 

Considerando las características del problema, nuestras restricciones son: 

1. Condiciones iniciales:

$$
T_0 = t_0, \qquad V_0=v_0 \qquad L_{e,0}=l_{e,0}
$$

Estas ecuaciones establecen el estado inicial del sistema al comienzo del horizonte de planificación. Específicamente, fijan el número inicial de terneras, vaconas y vacas lecheras de cada edad de acuerdo con la información disponible de la granja.

2. Transición de terneras a vaconas:

$$
(1 -  \alpha_t) \cdot T_{y-1} - 1 \leq V_y \leq (1 -  \alpha_t) \cdot T_{y-1} + 1, \qquad \forall y \in Y 
$$

Esta familia de restricciones representan el paso de las terneras al grupo de vaconas entre dos años consecutivos. El número de vaconas depende del número de terneras existentes el año anterior y de la tasa de supervivencia correspondiente. Así, se garantiza una transición biológicamente consistente entre ambas etapas del desarrollo, evitando que el modelo genere vaconas sin provenir de terneras previamente existentes.

3. Transición de vaconas a vacas lecheras:

$$
(1 - \alpha_v) \cdot V_{y-1} - 1 \leq L_{2,y} \leq (1 - \alpha_v) \cdot V_{y-1} + 1, \qquad \forall y \in Y
$$

Esta familia de restricciones asegura que las vaconas que sobreviven pasan a formar parte del grupo de vacas lecheras de dos años, incorporándose así al sistema de producción de leche.

4. Envejecimiento de vacas lecheras:

$$
(1 - \alpha_e) \cdot  L_{e,y-1} - 1 \leq L_{e+1, y} \leq (1 - \alpha_e) \cdot  L_{e,y-1} + 1, \qquad \forall y \in Y, \forall e \in A_l \setminus \{m\}
$$

Esta familia de restricciones modela el envejecimiento anual de las vacas lecheras. Las vacas de edad $e$ que sobreviven pasan a la categoría de edad $e+1$, considerando una tasa de mortalidad específica para cada edad, con lo cual se representa de forma más realista la mortalidad, evitando concentrar todas las pérdidas en un único grupo de edad.

En las últimas cuatro familias de restricciones, el término $\pm 1$ se introduce para absorber los efectos del redondeo derivados de trabajar con variables enteras y tasas de supervivencia expresadas como porcentajes. De esta manera, se restringe la solución a valores enteros cercanos al número esperado de animales supervivientes, evitando tanto la infactibilidad del modelo como desviaciones excesivas respecto al comportamiento biológico esperado. Se emplea $+1$ y $-1$ a cada lado de la expresión para darle mayor flexibilidad al modelo. 

5. Nacimiento de terneros:

$$
 M_y \leq \delta \cdot (1 - \beta) \cdot \sum_{e \in A_l} L_{e,y} , \qquad \forall y \in Y 
$$

Esta familia de restricciones asegura que la cantidad de terneros disponibles para la venta esté limitada por el número esperado de nacimientos masculino, es decir, impide que el modelo considere ventas superiores a los nacimientos biológicamente posibles. 

6. Nacimiento de terneras:

$$
T_y + H_y  \leq  \delta \cdot \beta \cdot \sum_{e \in A_l} L_{e,y} , \qquad \forall y \in Y
$$

Esta familia de restricciones permite que el modelo decida la mejor estrategia entre conservar o vender terneras, respetando la disponibilidad biológica de nacimientos. 


7. Capacidad del terreno:

$$
h_t \cdot T_y + h_v \cdot V_y + h_l \cdot \sum_{e \in A_l} L_{e,y} \leq H \qquad y \in Y
$$

Esta familia de restricciones asegura que el terreno para pastoreo empleado por el ganado conservado no supere la extensión del terreno total de la granja. 

8. Capacidad de la mano de obra y horas de trabajo requerido: 

$$
w_t \cdot T_y + w_v \cdot V_y + w_l \cdot \sum_{e \in A_l} L_{e,y} \leq S_y \cdot s + O_y \qquad \forall y \in Y
$$

Esta familia de restricciones garantiza que la carga laboral generada por el manejo del ganado deba ser cubierta mediante las horas ordinarias proporcionadas por los trabajadores contratados y, cuando sea necesario, mediante horas extraordinarias.

9. Requerimiento de la cantidad mínima de alimento:

$$
 r_t \cdot T_y + r_v \cdot V_y + r_l \cdot \sum_{e \in A_l} L_{e,y} \leq \sum_{x \in X} F_{x,y} \qquad \forall y \in Y
$$

Esta familia de restricciones asegura que la cantidad total de alimento adquirida debe ser suficiente para satisfacer los requerimientos nutricionales de todas las categorías de animales.

10. Capacidad de almacenamiento:

$$
F_{x,y} \leq   k \cdot b \qquad \forall x \in X, \forall y \in Y
$$

Esta familia de restricciones evita que el modelo proponga compras superiores a la capacidad física de almacenamiento, haciendo factibles las decisiones de abastecimiento. Se realiza mes a mes pues es la frecuencia en la que la granja se abastece de alimento.

11. Sostenibilidad: 

$$
T_{y+1} \geq L_{m,y}, \qquad \forall y \in Y \setminus \{n\}
$$

Estas restricciones favorecen el reemplazo gradual de los animales descartados y contribuyen a mantener la capacidad productiva del hato durante el horizonte de planificación.

12. Número mínimo de ganado al final del horizonte de tiempo

$$
Q \leq T_n + V_n + \sum_{e \in Al} L_{e,n}
$$

Finalmente, esta restricción genera que al finalizar el horizonte de planificación, el tamaño total del hato debe ser al menos igual a un nivel mínimo previamente establecido, lo cual evita que la optimización favorezca únicamente beneficios de corto plazo mediante la venta excesiva de animales y garantiza que la explotación conserve un inventario suficiente para continuar operando una vez concluido el horizonte de estudio.

### INSTANCIA 

```python
#edad a la que una vaca se vuelve de descarte 
maxe=11
#Horizonte de tiempo 
N=5 

#CONJUNTOS
Y=[i+1 for i in range(N)] #años
A=[i+1 for i in range(maxe)] #edades
Al=[i+2 for i in range(10)] #edades vaca adulta
X=[i +1 for i in range(12)] #meses 

#PARÁMETROS

#ganado
t0=8 #número inicial de terneras 
v0=10 #numero inicial de vaconas 
le0={2: 9, 3: 6, 4: 10, 5: 7, 6: 7, 7: 7, 8: 10, 9: 9, 10: 6, 11: 6} #número de vacas lecheras de cada edad
alpha_t=0.05 #mortalidad de terneras 
alpha_v=0.03 #mortalidad de vaconas
alpha_e = {2: 0.05, 3: 0.05, 4: 0.02, 5: 0.02, 6: 0.04, 7: 0.04, 8: 0.05, 9: 0.06, 10: 0.07}  #mortalidad por edad de vacas lecheras
beta=0.5  #tasa de nacidos hembras, machos
delta=0.95 #número de nacimientos exitosos 

#producción
mu=18 #cantidad de litros de leche producidos por día por una vaca lechera 
d=310 #días productivos en el día 

#precios en  USD
p_mu=0.5 # precio de venta por litro de leche 
p_tm=40 #precio de venta por ternero macho 
p_th=100 #precio de venta por ternera hembra 
p_D=1000 #precio de venta por vaca de descarte 


#alimentacion, medido en Kg
c_f=47.5/1000 #costo por kilo de alimento suplementario
r_t=3150 #alimento suplementario requerido por año por cada ternero 
r_v=720 #alimiento suplementario requerido por año por cada vacona 
r_l=12120 #alimeto requerido por vaca lechera 
b=2 #número de bodegas 
k=48000 #capacidad mensual de alimento que cabe en una bodega 
c_a=0.00222 #costo en USD que cuesta almacenar cada kg de alimento

#terreno en hectareas 
HT=25 #extensión del terreno de la granja
h_t=0.04 #terreno requerido por ternero 
h_v=0.17 #terreno requerido por vacona
h_l=0.24 #terreno requerido por vaca lechera 

#mano de obra en horas
w_t=10 +20 +10 #horas de manejo, gestión y saneamiento requerido por ternera 
w_v= 20 +10 +5 #horas de manejo gestión y saneamiento requerido por vacona 
w_l=52 + 52+ 30 #horas de manejo gestión y saneamientorequerido por vaca lechera
c_reg=3 #costo por hora regular
c_ext=5.0 #costo por hora extra 
s=2920 #horas que aporta un trabajador por año 

#costos de manejo y cuidado 
c_t=5 #costo por ternera
c_v=3 #costo por vacona
c_l=20 #costo por vaca lechera

#mínimo de ganado al final del horizonte de tiempo 
Q = 90
```

```python
#FLUJO DE CAJA 

r = 0.10 #tasa de descuento
pi_l = p_mu*mu*d - c_l - (c_f + c_a)*r_l   # utilidad neta anual de una vaca lechera activa (sin mano de obra)
pi_v = -(c_v + (c_f + c_a)*r_v)            # costo neto anual de una vacona
f_t = (1 - alpha_t) / (1 + r)
f_v = (1 - alpha_v) / (1 + r)
f_e = {e: (1 - alpha_e[e]) / (1 + r) for e in alpha_e}   # e = 2,...,10

# Valor de una vaca de edad e, calculado hacia atrás desde e=11
V_cow = {11: pi_l + p_D}
for e in range(10, 1, -1):
    V_cow[e] = pi_l + f_e[e] * V_cow[e+1]

V_vacona = pi_v + f_v * V_cow[2]   # valor total de ser vacona desde ese punto en adelante

# Bonos terminales (a aplicar al año 5, una vez)
p_g_T = f_t * V_vacona
p_g_V = f_v * V_cow[2]
terminal_L = {e: f_e[e] * V_cow[e+1] for e in range(2, 11)} 
terminal_L[11] = 0.0  
```

```python
#MODELO
def ejecutar_modelo(mu, p_mu, c_f, HT, h_t, h_v, h_l): 

    m=gp.Model("Planificación: Granja Agroecológica Camino al Sol")

    #VARIABLES

    T=m.addVars(Y, lb= 0, vtype=GRB.INTEGER, name="terneras")
    V=m.addVars(Y, lb= 0, vtype=GRB.INTEGER, name="vaconas")
    L=m.addVars(Al, Y, lb= 0, vtype=GRB.INTEGER, name="vacas_lecheras" )
    M = m.addVars(Y, lb= 0, vtype=GRB.INTEGER, name="venta_terneros") 
    H = m.addVars(Y, lb= 0, vtype=GRB.INTEGER, name="venta_terneras")
    F = m.addVars(X, Y, lb=0, vtype=GRB.CONTINUOUS, name="alimento_comprado")
    O = m.addVars(Y, vtype=GRB.INTEGER, name="horas_extra")
    S = m.addVars(Y,lb= 0, vtype=GRB.INTEGER, name="trabajadores")

    #RESTRICCIONES

    #primer año
    #vaconas
    m.addConstr(V[1] <= (1 - alpha_t)*t0 +1, "minimo_vaconas_año1")
    m.addConstr(V[1] >= (1 - alpha_t)*t0 - 1, "maximo_vaconas_año1")

    #vacas lecheras de edad 2
    m.addConstr(L[2,1] <= (1 -alpha_v) * v0 + 1, "vacas_edad2_año1")
    m.addConstr(L[2,1] >= (1 - alpha_v)*v0 - 1, "mínimo_vacas_edad2_año1")

    #vacas lecheras de edad mayor a 2 
    m.addConstrs((L[e+1,1] <= (1-alpha_e[e]) * le0[e] + 1 for e in Al if e!= maxe),"supervivencia_sup_año1")
    m.addConstrs((L[e+1,1] >= (1-alpha_e[e]) * le0[e] - 1 for e in Al if e!= maxe), "supervivencia_inf_año1")

    #transición de terneras a vaconas
    m.addConstrs((V[y] <= (1 - alpha_t) * T[y-1] + 1 for y in Y if y !=1), "envejecimiento_terneros")
    m.addConstrs((V[y] >= (1 - alpha_t)*T[y-1] - 1 for y in Y if y !=1), "minimo_vaconas")

    #transición a vaconas de edad 2 
    m.addConstrs((L[2,y] <= (1 - alpha_v) * V[y-1] + 1 for y in Y if y !=1), "envejecimiento_vaconas")
    m.addConstrs((L[2,y] >= (1 - alpha_v)*V[y-1] - 1 for y in Y if y !=1), "minimo_lecheras_edad2")

    #envejecimiento de vacas lecheras
    m.addConstrs((L[e+1,y] <= (1-alpha_e[e]) * L[e,y-1] + 1 for e in Al if e != maxe for y in Y if y!=1), "supervivencia_sup")
    m.addConstrs((L[e+1,y] >= (1-alpha_e[e]) * L[e,y-1] - 1 for e in Al if e != maxe for y in Y if y!=1), "supervivencia_inf")

    #nacimientos y venta de terneros y terneras
    m.addConstrs((M[y] <= delta * (1 - beta) * L.sum('*', y) for y in Y) , "machos_nacidos_sup")
    m.addConstrs((T[y] + H[y] <= delta * beta * L.sum('*', y) for y in Y), "hembras_nacidas_sup")

    #sostenibilidad 
    m.addConstrs((T[y+1] >= L[maxe,y] for y in Y if y!= Y[-1]), "sostenibilidad")

    #uso del terreno
    m.addConstrs((h_t * T[y] + h_v * V[y] + h_l * L.sum('*', y) <= HT for y in Y), "uso_terreno")

    #mano de obra 
    m.addConstrs((w_t * T[y] +w_v * V[y] + w_l * L.sum('*', y)  <= S[y] * s + O[y] for y in Y), "mano_obra_año")

    #compra de alimentos 
    m.addConstrs((r_t*T[y] + r_v * V[y] + r_l * L.sum('*', y) <= F.sum('*', y) for y in Y), "alimento_comprado")

    #almacenamiento
    m.addConstrs((F[x,y] <= k* b for x in X for y in Y), "almacenamiento_año")

    #mínimo de vacas al final del año 5
    m.addConstr(Q <= T[Y[-1]] + V[Y[-1]] + L.sum('*',Y[-1]), "vacas_final_año5")


    #FUNCIÓN OBJETIVO 
    Y_last = Y[-1]

    m.setObjective(
        gp.quicksum(
            p_mu*mu*d*L.sum('*', y) + p_tm*M[y] + p_th*H[y] + p_D*L[maxe,y]
            - (c_f + c_a)*F.sum('*', y) - c_reg*s*S[y] - c_ext*O[y]
            - (c_t*T[y] + c_v*V[y] + c_l*L.sum('*', y))
            for y in Y
        )
        + p_g_T * T[Y_last]
        + p_g_V * V[Y_last]
        + gp.quicksum(terminal_L[e] * L[e, Y_last] for e in Al),
        GRB.MAXIMIZE
    )

    m.optimize()

    if m.Status != GRB.OPTIMAL:
        print("El modelo no encontró una solución óptima.")
        return None
    
    return (m, T, M, H, V, L, S, O, F, mu, p_mu, c_f, HT, h_t, h_v, h_l)
```

```python
#TABLAS
def mostrar_tablas(m, T, M, H, V, L, S, O, F, mu, p_mu, c_f, HT, h_t, h_v, h_l):
    if m.Status != GRB.OPTIMAL:
        print("El modelo no encontró una solución óptima.")
        return

    print(f"\nUtilidad total máxima: ${m.ObjVal:,.2f}")
    resumen = {}

    resumen["Año 0"] = {

        "Terneras": t0,
        "Vaconas": v0,
        "Vacas adultas": sum(le0.values()),
        "Terneros vendidos": None,
        "Terneras vendidas": None,
        "Vacas descarte": le0[maxe],
        "Alimento comprado (kg)": None,
        "Alimento consumido (kg)":
            r_t*t0 + r_v*v0 + r_l*sum(le0.values()),

        "Trabajadores": None,
        "Horas extra": None,
        "Horas utilizadas":
            w_t*t0 + w_v*v0 + w_l*sum(le0.values()),

        "Hectáreas utilizadas":
            h_t*t0 + h_v*v0 + h_l*sum(le0.values()),
    }

    for y in Y:

        total_adultas = sum(L[e,y].X for e in Al)
        alimento_cons = (r_t*T[y].X + r_v*V[y].X + r_l*total_adultas)
        horas_usadas = ( w_t*T[y].X + w_v*V[y].X + w_l*total_adultas)
        hectareas = ( h_t*T[y].X + h_v*V[y].X + h_l*total_adultas)
        utilidad = (
            p_mu*mu*d*total_adultas
            + p_tm*M[y].X + p_th*H[y].X + p_D*L[maxe,y].X
            - (c_f+c_a)*sum(F[x,y].X for x in X)
            - c_reg*s*S[y].X - c_ext*O[y].X
            - (c_t*T[y].X + c_v*V[y].X + c_l*total_adultas)
        )

        resumen[f"Año {y}"] = {

            "Terneras": int(round(T[y].X)),
            "Vaconas": int(round(V[y].X)),
            "Vacas adultas": int(round(total_adultas)),
            "Terneros vendidos": int(round(M[y].X)),
            "Terneras vendidas": int(round(H[y].X)),
            "Vacas descarte": int(round(L[maxe,y].X)),
            "Alimento comprado (kg)":
                round(sum(F[x,y].X for x in X),2),
            "Alimento consumido (kg)":
                round(alimento_cons,2),
            "Trabajadores":
                int(round(S[y].X)),
            "Horas extra":
                round(O[y].X,2),
            "Horas utilizadas":
                round(horas_usadas,2),
            "Hectáreas utilizadas":
                round(hectareas,2),
            "Utilidad real":
                round(utilidad,2)
        }

    df_resumen = pd.DataFrame(resumen)

    print("\n================================== TABLAS DE RESUMEN ===============================\n")

    display(
        df_resumen.fillna("-").rename_axis("Variable")
    )


    hato = {}

    hato["Año 0"] = {

        "Terneras": t0,
        "Vaconas": v0,
        **{
            f"Vacas {e} años": le0[e]
            for e in Al
        }
    }

    for y in Y:

        hato[f"Año {y}"] = {

            "Terneras": int(round(T[y].X)),
            "Vaconas": int(round(V[y].X)),
            **{
                f"Vacas {e} años":
                int(round(L[e,y].X))

                for e in Al
            }
        }

    df_hato = pd.DataFrame(hato)

    display(
        df_hato.rename_axis("Categoría")
    )
```

```python
mostrar_tablas(m, T, M, H, V, L, S, O, F, leche_prod1, p_leche1, costo_comida1, HT1, h_t1, h_v1, h_l1)
```

## GRÁFICAS

```python
def mostrar_graficas(m, T, M, H, V, L, S, O, F,leche_prod,  precio_leche,  costo_comida, HT, h_t, h_v, h_l):

    if m.Status != GRB.OPTIMAL:
        print("El modelo no encontró una solución óptima.")
        return

    # ==================  1. COMPOSICIÓN DEL HATO  ==================

    datos_hato = []

    datos_hato.append({"Año": "Año 0", "Categoría": "Terneras", "Cantidad": t0})
    datos_hato.append({"Año": "Año 0", "Categoría": "Vaconas", "Cantidad": v0})

    for e in Al:
        datos_hato.append({"Año": "Año 0", "Categoría": f"Vacas {e} años", "Cantidad": le0[e]})

    for y in Y:

        datos_hato.append({"Año": f"Año {y}", "Categoría": "Terneras", "Cantidad": T[y].X})
        datos_hato.append({"Año": f"Año {y}", "Categoría": "Vaconas", "Cantidad": V[y].X})

        for e in Al:
            datos_hato.append({"Año": f"Año {y}", "Categoría": f"Vacas {e} años", "Cantidad": L[e, y].X})

    df = pd.DataFrame(datos_hato)
    fig = px.bar(df, x="Año", y="Cantidad", color="Categoría", title="Composición del hato por categoría y edad")
    fig.update_traces(width=0.8)
    fig.update_layout(width=700, height=600, bargap=0.4)
    fig.show()

    # ==================  2. UTILIDAD  ==================

    utilidad = []

    for y in Y:

        total_adultas = L.sum('*', y).getValue()

        u = (
            precio_leche * leche_prod * d * total_adultas
            + p_tm * M[y].X + p_th * H[y].X + p_D * L[maxe, y].X
            - (costo_comida + c_a) * F.sum('*', y).getValue()
            - c_reg * s * S[y].X - c_ext * O[y].X
            - (c_t * T[y].X + c_v * V[y].X + c_l * total_adultas)
        )

        utilidad.append({"Año": f"Año {y}", "Utilidad": u})

    df = pd.DataFrame(utilidad)
    fig = px.line(df, x="Año", y="Utilidad", title="Utilidad anual",
                  text=df["Utilidad"].apply(lambda v: f"{v:,.2f}"),
                  markers=True,
                  color_discrete_sequence=["red"])
    fig.update_traces(textposition="top center")
    fig.update_layout(width=700, height=500)
    fig.show()

    # ==================  3. EGRESOS  ==================

    costos = []

    for y in Y:

        alimento = (costo_comida + c_a) * F.sum('*', y).getValue()
        trabajadores = c_reg * s * S[y].X
        extras = c_ext * O[y].X
        mantenimiento = (c_t * T[y].X + c_v * V[y].X + c_l * L.sum('*', y).getValue())
        total = alimento + trabajadores + extras + mantenimiento

        costos.extend([
            {"Año": f"Año {y}", "Concepto": "Alimento", "Porcentaje": 100 * alimento / total},
            {"Año": f"Año {y}", "Concepto": "Trabajadores", "Porcentaje": 100 * trabajadores / total},
            {"Año": f"Año {y}", "Concepto": "Horas extra", "Porcentaje": 100 * extras / total},
            {"Año": f"Año {y}", "Concepto": "Mantenimiento", "Porcentaje": 100 * mantenimiento / total},
        ])

    df = pd.DataFrame(costos)

    fig = px.bar(df, x="Año", y="Porcentaje", color="Concepto",
                 title="Composición porcentual de egresos", text=df["Porcentaje"].round(1))
    fig.update_traces(width=0.8)
    fig.update_layout(yaxis_title="% de egresos", width=700, height=600, bargap=0.4)
    fig.show()

    # ==================  4. UTILIZACIÓN DE RECURSOS  ==================

    recursos = []

    for y in Y:

        recursos.extend([
            {
                "Año": f"Año {y}",
                "Recurso": "Terreno",
                "Utilización": 100 * (
                    h_t * T[y].X + h_v * V[y].X + h_l * L.sum('*', y).getValue()
                ) / HT,
            },
            {
                "Año": f"Año {y}",
                "Recurso": "Mano de obra",
                "Utilización": 100 * (
                    w_t * T[y].X + w_v * V[y].X + w_l * L.sum('*', y).getValue()
                ) / (S[y].X * s + O[y].X),
            },
            {
                "Año": f"Año {y}",
                "Recurso": "Almacenamiento",
                "Utilización": 100 * F.sum('*', y).getValue() / (12 * k * b),
            },
        ])

    df = pd.DataFrame(recursos)

    fig = px.line(
        df,
        x="Año",
        y="Utilización",
        color="Recurso",
        markers=True,
        title="Porcentaje de utilización de recursos",
    )
    fig.update_layout(yaxis_title="% de utilización", width=700, height=400)
    fig.show()
```

```python
mostrar_graficas(m, T, M, H, V, L, S, O, F, leche_prod1, p_leche1, costo_comida1, HT1, h_t1, h_v1, h_l1)
```
